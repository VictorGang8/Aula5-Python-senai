1-Crie um programa que solicite ao usuário duas idades e, em seguida, exiba as
comparações entre elas.

25 é igual a 30?

25 é diferente de 30?

25 é maior que 30?

25 é maior ou igual a 30?

25 é menor que 30?

25 é menor ou igual a 30?

```python
idade1 = int(input('Digite a idade da primeira pessoa: '))
idade2 = int(input('Digite a idade da segunda pessoa: '))
print(idade1, 'é igual a', idade2, '?:', idade1 == idade2)
print(idade1, 'é diferente de', idade2, '?:', idade1 != idade2)
print(idade1, 'é maior que', idade2, '?:', idade1 > idade2)
print(idade1, 'é maior ou igual a', idade2, '?:', idade1 >= idade2)
print(idade1, 'é menor que', idade2, '?:', idade1 < idade2)
print(idade1, 'é menor ou igual a', idade2, '?:', idade1 <= idade2)
```

Atividade 4 – Promoção na Loja

Uma loja está fazendo uma promoção.

O cliente ganha desconto se:

For cliente VIP OU

Tiver feito uma compra acima de R$100

O programa deve:

Pedir:

Se o cliente é VIP (sim ou nao)

O valor da compra

Mostrar o resultado da verificação (True ou False)

```python
vip = input('O cliente é VIP?(sim/não) ')
compra = float(input('Digite o valor da compra: '))
resultado = (vip == 'sim') or (compra > 100)
if resultado == True:
  print('O cliente ganhou desconto')
else:
  print('O cliente não ganhou desconto')
print(resultado)
```

Atividade 5– Dia de descanso

Peça:

Dia da semana

Considere que:

Dias de descanso: sabado ou domingo

Mostre o resultado da verificação:
A pessoa está de descanso se NÃO for dia útil (segunda a sexta)

```python
dia = input('Digite o dia da semana: ')
resultado = (dia != 'segunda') and (dia != 'terça') and (dia != 'quarta') and (dia != 'quinta') and (dia != 'sexta')
if resultado == True:
  print('A pessoa está de descanso')
else:
  print('A pessoa não está de descanso')


print(resultado)
```
